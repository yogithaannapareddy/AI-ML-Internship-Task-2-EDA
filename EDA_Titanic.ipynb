```python
numeric_columns = ['Age', 'Fare', 'SibSp', 'Parch']

print("Mean:")
print(df[numeric_columns].mean())

print("\nMedian:")
print(df[numeric_columns].median())

print("\nStandard Deviation:")
print(df[numeric_columns].std())
```

```python
numeric_columns = ['Age', 'Fare', 'SibSp', 'Parch']

for column in numeric_columns:
    plt.figure(figsize=(8, 5))
    
    plt.hist(df[column].dropna(), bins=30)
    
    plt.title(f'Distribution of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    
    plt.show()
```

```python
numeric_columns = ['Age', 'Fare', 'SibSp', 'Parch']

for column in numeric_columns:
    plt.figure(figsize=(8, 5))
    
    sns.boxplot(x=df[column])
    
    plt.title(f'Boxplot of {column}')
    plt.xlabel(column)
    
    plt.show()
```

```python
# Select numerical columns
numeric_columns = ['PassengerId', 'Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']

# Calculate correlation
correlation_matrix = df[numeric_columns].corr()

# Display correlation values
correlation_matrix
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix of Titanic Dataset')
plt.show()
```

```python
pairplot_columns = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']

sns.pairplot(
    df[pairplot_columns].dropna(),
    hue='Survived'
)

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Survival by Gender
sns.countplot(data=df, x='Sex', hue='Survived', ax=axes[0])
axes[0].set_title('Survival Count by Gender')
axes[0].set_xlabel('Gender')
axes[0].set_ylabel('Number of Passengers')

# Survival by Passenger Class
sns.countplot(data=df, x='Pclass', hue='Survived', ax=axes[1])
axes[1].set_title('Survival Count by Passenger Class')
axes[1].set_xlabel('Passenger Class')
axes[1].set_ylabel('Number of Passengers')

plt.tight_layout()
plt.show()
```

```python
print("Survival Percentage by Gender:")
print(df.groupby('Sex')['Survived'].mean() * 100)

print("\nSurvival Percentage by Passenger Class:")
print(df.groupby('Pclass')['Survived'].mean() * 100)
```

```python
numeric_columns = ['Age', 'Fare', 'SibSp', 'Parch']

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"{column}:")
    print(f"  Lower Bound: {lower_bound:.2f}")
    print(f"  Upper Bound: {upper_bound:.2f}")
    print(f"  Number of Outliers: {len(outliers)}")
    print()
```

```python
print("Skewness of Numerical Features:")
print(df[numeric_columns].skew())
```
